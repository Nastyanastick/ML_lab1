2. Нормализация данных

```python
def z_score_normalization(data):
    mean = np.mean(data, axis=0)
    std = np.std(data, axis=0)
    normalized_data = (data - mean) / std
    return normalized_data
```

```python
age_normalized = z_score_normalization(train['Age'].dropna())
print('Среднее после нормализации: ', np.mean(age_normalized))
print('Стандартное отклонение после нормализации: ', np.std(age_normalized))
```

```python
def min_max_normalization(data):
    min_value = np.min(data)
    max_value = np.max(data)
    normalized_data = (data - min_value) / (max_value - min_value)
    return normalized_data
```

```python
age_normalized = min_max_normalization(train['Age'].dropna())

print('Минимум:', np.min(age_normalized))
print('Максимум:', np.max(age_normalized))
print('Среднее после нормализации: ', np.mean(age_normalized))
print('Стандартное отклонение после нормализации: ', np.std(age_normalized))
```

3. Линейная регрессия

```python
class LinearRegressionCustom:
    def __init__(self, learning_rate=0.01, n_iterations=1000):
        self.weights = None
        self.bias = None
        self.learning_rate = learning_rate
        self.n_iterations = n_iterations


    def fit(self, X, y, method='analytic', regularization=None, alpha=0.01, l1_ratio=0.5, p=2):
        X = np.asarray(X)
        y = np.asarray(y)
        X_with_bias = np.column_stack((np.ones(X.shape[0]), X))

        if alpha < 0:
            raise ValueError("alpha не может быть отрицательным")

        if not 0 <= l1_ratio <= 1:
            raise ValueError("l1_ratio должен быть от 0 до 1")

        if regularization not in [None, 'l1', 'l2', 'elastic_net', 'lp']:
            raise ValueError("Неизвестный тип регуляризации")

        if method == 'analytic' and regularization is not None:
            raise ValueError("Регуляризация доступна только для gd и sgd")
        if regularization == 'lp' and p < 1:
            raise ValueError("Для Lp-регуляризации p должно быть не меньше 1")

        if method == 'analytic':
            params = np.linalg.pinv(X_with_bias) @ y
            self.bias = params[0]
            self.weights = params[1:]
        
        elif method == 'gd':
            self.weights = np.zeros(X.shape[1])
            self.bias = 0
            for _ in range(self.n_iterations):
                y_pred = X @ self.weights + self.bias
                error = y_pred - y
                dw = (2 / len(y)) * (X.T @ error)
                db = (2 / len(y)) * np.sum(error)

                if regularization == 'l1':
                    dw += alpha * np.sign(self.weights)
                elif regularization == 'l2':
                    dw += 2 * alpha * self.weights
                elif regularization == 'elastic_net':
                    dw += alpha * (l1_ratio * np.sign(self.weights) + 2 * (1 - l1_ratio) * self.weights)
                elif regularization == 'lp':
                    dw += alpha * p * np.sign(self.weights) * np.abs(self.weights) ** (p - 1)
                self.weights -= self.learning_rate * dw
                self.bias -= self.learning_rate * db

        elif method == 'sgd':
            self.weights = np.zeros(X.shape[1])
            self.bias = 0
            for _ in range(self.n_iterations):
                indexes = np.random.permutation(len(y))
                for i in indexes:
                    xi = X[i]
                    yi = y[i]
                    y_pred = xi @ self.weights + self.bias
                    error = y_pred - yi
                    dw = 2 * xi * error
                    db = 2 * error

                    if regularization == 'l1':
                        dw += alpha * np.sign(self.weights)
                    elif regularization == 'l2':
                        dw += 2 * alpha * self.weights
                    elif regularization == 'elastic_net':
                        dw += alpha * (l1_ratio * np.sign(self.weights) + 2 * (1 - l1_ratio) * self.weights)
                    elif regularization == 'lp':
                        dw += alpha * p * np.sign(self.weights) * np.abs(self.weights) ** (p - 1)
                    
                    self.weights -= self.learning_rate * dw
                    self.bias -= self.learning_rate * db
        else:
            raise ValueError("Неизвестный метод обучения")

        return self


    def predict(self, X):
        X = np.asarray(X)
        return X @ self.weights + self.bias
```

5. метрика MSE

```python
def mse (y_true, y_pred):
    y_true = np.asarray(y_true)
    y_pred = np.asarray(y_pred)
    return np.mean((y_true - y_pred) ** 2)
```

```python
from sklearn.model_selection import train_test_split, KFold, cross_val_score
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder, StandardScaler, SplineTransformer, PolynomialFeatures
from sklearn.base import clone

valid_target = train['RiskScore'].notna() & ~train['RiskScore'].isin([-9999999, 10000000])
train_labeled = train.loc[valid_target].copy()

train_model_data = train_labeled.dropna().copy()
y = train_model_data['RiskScore'].to_numpy(dtype=float)

print('Строк в исходном train:', len(train))
print('Строк с корректным RiskScore:', len(train_labeled))
print('Строк без пропусков в признаках:', len(train_model_data))
print('Исключено строк из-за пропусков в признаках:', len(train_labeled) - len(train_model_data))

def make_features(data):
    X = data.drop(columns=['RiskScore', 'ID'], errors='ignore').copy()

    dates = pd.to_datetime(X.pop('ApplicationDate'), errors='coerce')
    X['ApplicationYear'] = dates.dt.year
    X['ApplicationMonth'] = dates.dt.month
    X['LoanToIncome'] = X['LoanAmount'] / X['AnnualIncome'].clip(lower=1)
    X['DebtPaymentsToIncome'] = X['MonthlyDebtPayments'] / X['MonthlyIncome'].clip(lower=1)
    X['LoanPaymentToIncome'] = X['MonthlyLoanPayment'] / X['MonthlyIncome'].clip(lower=1)
    X['NetWorthToIncome'] = X['NetWorth'] / X['AnnualIncome'].clip(lower=1)
    X['AssetsToIncome'] = X['TotalAssets'] / X['AnnualIncome'].clip(lower=1)
    X['LiabilitiesToIncome'] = X['TotalLiabilities'] / X['AnnualIncome'].clip(lower=1)
    X['AgeExperienceGap'] = X['Age'] - X['Experience']

    return X

X_original = make_features(train_model_data)
X_train_df, X_val_df, y_train_raw, y_val_raw = train_test_split(
    X_original, y, test_size=0.2, random_state=42
)

numeric_features = X_train_df.select_dtypes(include=np.number).columns.tolist()
categorical_features = X_train_df.select_dtypes(exclude=np.number).columns.tolist()

poly_features = [
    'CreditScore', 'MonthlyIncome', 'BaseInterestRate', 'AnnualIncome',
    'InterestRate', 'TotalDebtToIncomeRatio', 'NetWorth', 'Age'
]

numeric_spline_pipeline = Pipeline([
    ('imputer', SimpleImputer(strategy='median')),
    ('spline', SplineTransformer(n_knots=6, degree=3, knots='quantile', include_bias=False)),
    ('scaler', StandardScaler())
])

polynomial_pipeline = Pipeline([
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler_before_poly', StandardScaler()),
    ('poly', PolynomialFeatures(degree=3, include_bias=False)),
    ('scaler_after_poly', StandardScaler())
])

try:
    one_hot_encoder = OneHotEncoder(handle_unknown='ignore', sparse_output=False)
except TypeError:
    one_hot_encoder = OneHotEncoder(handle_unknown='ignore', sparse=False)

categorical_pipeline = Pipeline([
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('onehot', one_hot_encoder)
])

preprocessor = ColumnTransformer([
    ('spline', numeric_spline_pipeline, numeric_features),
    ('polynomial', polynomial_pipeline, poly_features),
    ('categorical', categorical_pipeline, categorical_features)
], sparse_threshold=0)

X_train_transformed = preprocessor.fit_transform(X_train_df)
X_val_transformed = preprocessor.transform(X_val_df)

print('Размер матрицы после преобразования:', X_train_transformed.shape)
print('Размер валидационной матрицы:', X_val_transformed.shape)
print('Есть NaN после преобразования:', np.isnan(X_train_transformed).any())
```

```python
from scipy.special import ndtr

def make_probability_features(data):
    margin = data['ApprovalScore'].to_numpy(dtype=float)
    weighted = data['BaseRiskScore'].to_numpy(dtype=float)

    features = {}

    for threshold in [0.8, 1.0, 1.2]:
        for sigma in [0.1, 0.25, 0.5, 1.0]:

            probability = ndtr((threshold - margin) / sigma)

            name = f'{threshold}_{sigma}'.replace('.', '_')

            features[f'ApprovalProb_{name}'] = probability
            features[f'BaseRiskProb_{name}'] = weighted * probability

    return pd.DataFrame(features, index=data.index)
```

```python

def add_risk_features(data):
    X = data.copy()

    X['RiskCredit'] = np.select(
        [X['CreditScore'] >= 750, X['CreditScore'] >= 700,
         X['CreditScore'] >= 650, X['CreditScore'] >= 600],
        [1, 2, 3, 4], default=5
    )

    X['RiskDTI'] = np.select(
        [X['DebtToIncomeRatio'] < 0.20, X['DebtToIncomeRatio'] < 0.30,
         X['DebtToIncomeRatio'] < 0.40, X['DebtToIncomeRatio'] < 0.50],
        [1, 2, 3, 4], default=5
    )

    X['RiskPaymentHistory'] = np.select(
        [X['PaymentHistory'] >= 99, X['PaymentHistory'] >= 97,
         X['PaymentHistory'] >= 95, X['PaymentHistory'] >= 90],
        [1, 2, 3, 4], default=5
    )

    X['RiskBankruptcy'] = np.where(X['BankruptcyHistory'] == 1, 5, 1)

    X['RiskPreviousDefaults'] = np.where(
        X['PreviousLoanDefaults'] == 0, 1,
        np.where(X['PreviousLoanDefaults'] == 1, 3, 5)
    )

    X['RiskUtilization'] = np.select(
        [X['CreditCardUtilizationRate'] < 0.20,
         X['CreditCardUtilizationRate'] < 0.40,
         X['CreditCardUtilizationRate'] < 0.60,
         X['CreditCardUtilizationRate'] < 0.80],
        [1, 2, 3, 4], default=5
    )

    X['RiskCreditHistory'] = np.select(
        [X['LengthOfCreditHistory'] >= 10,
         X['LengthOfCreditHistory'] >= 7,
         X['LengthOfCreditHistory'] >= 5,
         X['LengthOfCreditHistory'] >= 3],
        [1, 2, 3, 4], default=5
    )

    X['RiskIncome'] = np.select(
        [X['AnnualIncome'] >= 120000, X['AnnualIncome'] >= 80000,
         X['AnnualIncome'] >= 50000, X['AnnualIncome'] >= 30000],
        [1, 2, 3, 4], default=5
    )

    X['RiskEmployment'] = np.select(
        [X['EmploymentStatus'] == 'Employed',
         X['EmploymentStatus'] == 'Self-Employed',
         X['EmploymentStatus'] == 'Unemployed'],
        [1, 2, 4], default=4
    )

    X['RiskNetWorth'] = np.select(
        [X['NetWorth'] >= 500000, X['NetWorth'] >= 250000,
         X['NetWorth'] >= 100000, X['NetWorth'] >= 50000],
        [1, 2, 3, 4], default=5
    )

    X['BaseRiskScore'] = (
        X['RiskCredit'] * 3 +
        X['RiskDTI'] * 2 +
        X['RiskPaymentHistory'] * 2 +
        X['RiskBankruptcy'] * 3 +
        X['RiskPreviousDefaults'] * 3 +
        X['RiskUtilization'] +
        X['RiskCreditHistory'] +
        X['RiskIncome'] +
        X['RiskEmployment'] +
        X['RiskNetWorth'] * 2
    )

    education_score = X['EducationLevel'].map({
        'High School': 0.2,
        'Associate': 0.1,
        'Bachelor': 0,
        'Master': -0.1,
        'Doctorate': -0.2
    }).fillna(0)

    month = X['ApplicationMonth']

    X['ApprovalScore'] = (
        (X['CreditScore'] - 600) / 250 +
        (100000 - X['AnnualIncome']) / 100000 +
        (X['TotalDebtToIncomeRatio'] - 0.4) * 2 +
        (X['LoanAmount'] - 10000) / 90000 +
        (X['InterestRate'] - 0.05) * 10 +
        0.5 * (X['BankruptcyHistory'] == 1) +
        0.3 * (X['PreviousLoanDefaults'] == 1) +
        0.2 * (X['EmploymentStatus'] == 'Unemployed') -
        0.1 * X['HomeOwnershipStatus'].isin(['Own', 'Mortgage']) -
        X['PaymentHistory'] / 120 -
        X['LengthOfCreditHistory'] / 60 -
        X['NetWorth'] / 500000 +
        np.abs(X['Age'] - 40) / 100 -
        X['Experience'] / 200 +
        education_score -
        0.1 * month.between(3, 8)
    )

    X['PredictedLoanApproved'] = (X['ApprovalScore'] < 1).astype(float)

    X['RiskScoreApprovedApprox'] = (
        X['BaseRiskScore'] *
        (1 - 0.2 * X['PredictedLoanApproved'])
    )

    return X

X_mse = add_risk_features(X_original)

X_train_mse_df, X_val_mse_df, y_train_mse_raw, y_val_mse_raw = train_test_split(
    X_mse, y, test_size=0.2, random_state=42
)

numeric_features_mse = X_train_mse_df.select_dtypes(include=np.number).columns.tolist()
categorical_features_mse = X_train_mse_df.select_dtypes(exclude=np.number).columns.tolist()

poly_features_mse = [
    'CreditScore', 'MonthlyIncome', 'BaseInterestRate', 'AnnualIncome',
    'InterestRate', 'TotalDebtToIncomeRatio', 'NetWorth', 'Age'
]

spline_features_mse = [
    'CreditScore', 'AnnualIncome', 'MonthlyIncome', 'BaseInterestRate',
    'InterestRate', 'TotalDebtToIncomeRatio', 'NetWorth', 'LoanAmount',
    'MonthlyLoanPayment', 'ApprovalScore', 'BaseRiskScore',
    'RiskScoreApprovedApprox', 'DebtToIncomeRatio', 'Age'
]

linear_features_mse = [
    col for col in numeric_features_mse
    if col not in spline_features_mse
]

numeric_spline_pipeline_mse = Pipeline([
    ('imputer', SimpleImputer(strategy='median')),
    ('spline', SplineTransformer(n_knots=6, degree=3, knots='quantile', include_bias=False)),
    ('scaler', StandardScaler())
])

numeric_linear_pipeline_mse = Pipeline([
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler())
])

polynomial_pipeline_mse = Pipeline([
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler_before_poly', StandardScaler()),
    ('poly', PolynomialFeatures(degree=2, include_bias=False)),
    ('scaler_after_poly', StandardScaler())
])

try:
    one_hot_encoder_mse = OneHotEncoder(handle_unknown='ignore', sparse_output=False)
except TypeError:
    one_hot_encoder_mse = OneHotEncoder(handle_unknown='ignore', sparse=False)

categorical_pipeline_mse = Pipeline([
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('onehot', one_hot_encoder_mse)
])

preprocessor_mse = ColumnTransformer([
    ('spline', numeric_spline_pipeline_mse, spline_features_mse),
    ('linear', numeric_linear_pipeline_mse, linear_features_mse),
    ('polynomial', polynomial_pipeline_mse, poly_features_mse),
    ('categorical', categorical_pipeline_mse, categorical_features_mse)
], sparse_threshold=0)

X_train_mse_transformed = preprocessor_mse.fit_transform(X_train_mse_df)
X_val_mse_transformed = preprocessor_mse.transform(X_val_mse_df)

P_train = make_probability_features(X_train_mse_df).to_numpy(dtype=float)
P_val = make_probability_features(X_val_mse_df).to_numpy(dtype=float)

prob_mean = P_train.mean(axis=0)
prob_std = P_train.std(axis=0)
prob_std[prob_std == 0] = 1
P_train = (P_train - prob_mean) / prob_std
P_val = (P_val - prob_mean) / prob_std
X_train_mse_transformed = np.column_stack([
    X_train_mse_transformed, P_train
])

X_val_mse_transformed = np.column_stack([
    X_val_mse_transformed, P_val
])

print('Размер новой матрицы:', X_train_mse_transformed.shape)
print('Размер проверочной матрицы:', X_val_mse_transformed.shape)
```

```python
from sklearn.metrics import mean_squared_error

model_validation = LinearRegressionCustom(
    learning_rate=0.01468,
    n_iterations=1000
)

model_validation.fit(
    X_train_transformed,
    y_train_raw,
    method='gd',
    regularization='l2',
    alpha=0.001
)

predictions_val = model_validation.predict(X_val_transformed)

custom_mse = mse(y_val_raw, predictions_val)
sklearn_mse = mean_squared_error(y_val_raw, predictions_val)

model_validation = LinearRegressionCustom(
    learning_rate=0.01,
    n_iterations=2500
)

model_validation.fit(
    X_train_mse_transformed,
    y_train_mse_raw,
    method='gd',
    regularization='l2',
    alpha=0.0005
)

predictions_val_mse = model_validation.predict(X_val_mse_transformed)

custom_mse = mse(y_val_mse_raw, predictions_val_mse)
sklearn_mse = mean_squared_error(y_val_mse_raw, predictions_val_mse)

print('Custom MSE:', custom_mse)
print('Sklearn MSE:', sklearn_mse)
print('MSE < 18:', custom_mse < 18)
```

4. кросс-валидация

```python
def k_fold_cross_validation(X, y, k, shuffle=False, random_state=None):
    X = np.asarray(X)
    y = np.asarray(y)
    n = len(X)
    if k < 2 or k > n:
        raise ValueError("k должно быть от 2 до количества объектов")
    indexes = np.arange(n)
    if shuffle:
        rng = np.random.RandomState(random_state)
        rng.shuffle(indexes)
    folds = np.array_split(indexes, k)
    scores = []
    for i in range(k):
        test_indexes = folds[i]
        train_indexes = np.concatenate(folds[:i] + folds[i + 1:])
        X_train, X_test = X[train_indexes], X[test_indexes]
        y_train, y_test = y[train_indexes], y[test_indexes]
        mean = np.mean(X_train, axis=0)
        std = np.std(X_train, axis=0)
        std[std == 0] = 1
        X_train = (X_train - mean) / std
        X_test = (X_test - mean) / std
        model = LinearRegressionCustom()
        model.fit(X_train, y_train, method='analytic')
        predictions = model.predict(X_test)
        score = mse(y_test, predictions)
        scores.append(score)
    return np.mean(scores), scores


def loo_cross_validation(X, y, shuffle=False, random_state=None):
    mean_score, scores = k_fold_cross_validation(X, y, k=len(X), shuffle=shuffle, random_state=random_state)
    return mean_score, scores
```

```python
from sklearn.model_selection import KFold, cross_val_score
from sklearn.linear_model import LinearRegression
from sklearn.pipeline import Pipeline
kf = KFold(n_splits=5, shuffle=True, random_state=42)
model = Pipeline([
    ('preprocessor', clone(preprocessor)),
    ('regression', LinearRegression())
])
scores = cross_val_score(
    model, X_original, y,
    cv=kf,
    scoring='neg_mean_squared_error'
)
print("Средний MSE sklearn:", -scores.mean())
```

7. R^2

```python
def r2_score(y_true, y_pred):
    y_true = np.asarray(y_true)
    y_pred = np.asarray(y_pred)
    ss_res = np.sum((y_true - y_pred) ** 2)
    ss_tot = np.sum((y_true - np.mean(y_true)) ** 2)
    if ss_tot == 0:
        raise ValueError("Невозможно вычислить R2: все реальные значения одинаковые")
    return 1 - ss_res / ss_tot
```

```python
from sklearn.metrics import r2_score as sklearn_r2
custom_r2 = r2_score(y_test, predictions)
sklearn_r2_value = sklearn_r2(y_test, predictions)
print('Custom R2:', custom_r2)
print('Sklearn R2:', sklearn_r2_value)
```

8. MAPE

```python
def mape(y_true, y_pred):
    y_true = np.asarray(y_true)
    y_pred = np.asarray(y_pred)
    if np.any(y_true == 0):
        raise ValueError("MAPE не определена для нулевых реальных значений")
    return np.mean(np.abs((y_true - y_pred) / y_true)) * 100
```
